```python
from ultralytics import YOLO

# load model (can be a yolov8 .pt or other compatible .pt)
model = YOLO("/content/ultralytics/runs/detect/train2/weights/best.pt")

# run a single image prediction and save results
results = model.predict(
    source="/content/ultralytics/Website-Screenshots-1/train/images/abc_net_au_png.rf.8b182077aea95ef51428b335c0593506.jpg",
    imgsz=640,
    save=True,            # save output images with boxes
    project="./runs/detect",
    name="exp_yolov8",
    device="cpu"          # or device=0 if GPU available
)

# show the prediction path and confidence for first image
print("Saved to:", results[0].path if len(results) else "no results")
print("Boxes:", results[0].boxes.xyxy, "Scores:", results[0].boxes.conf)
```

```python
from ultralytics import YOLO

# Load your trained model
model = YOLO("/content/ultralytics/runs/detect/train2/weights/best.pt")

# Validate on the custom dataset
metrics = model.val(data="custom_dataset.yaml", imgsz=640, batch=16, device="cpu")

# Print the results
print("Validation metrics:")
print(metrics)
```
